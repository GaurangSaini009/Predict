```python
def clean_data(value):
    value = value.split("")[0]
    value = value.strip()
    if value == "":
        value =0
        return float(value)
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
cars_data["fuel"]=le.fit_transform(cars_data["fuel"])
cars_data["seller_type"]=le.fit_transform(cars_data["seller_type"])
cars_data["owner"]=le.fit_transform(cars_data["owner"])
cars_data["name"]=le.fit_transform(cars_data["name"])
```

```python
cars_data.head()
```

```python
cars_data["transmission"]=le.fit_transform(cars_data["transmission"])
```

```python
input_data= cars_data.drop(columns=["selling_price"])
output_data=cars_data["selling_price"]
```

```python
cars_data.head()
```

```python
x_train,x_test, y_train, y_test = train_test_split(input_data,output_data,test_size=0.2)
```

```python
model= LinearRegression()
```

```python

if isinstance(x_train, pd.DataFrame):
    x_train_cleaned = x_train.copy()
    for col in x_train_cleaned.columns:
        if x_train_cleaned[col].dtype == 'object': 
            x_train_cleaned[col] = x_train_cleaned[col].str.replace(r'[^\d.]', '', regex=True)
            x_train_cleaned[col] = pd.to_numeric(x_train_cleaned[col], errors='coerce')
    
    if isinstance(y_train, pd.Series) and y_train.dtype == 'object':
        y_train_cleaned = y_train.str.replace(r'[^\d.]', '', regex=True)
        y_train_cleaned = pd.to_numeric(y_train_cleaned, errors='coerce')
    else:
        y_train_cleaned = y_train
else:
    x_train_cleaned = []
    for row in x_train:
        if isinstance(row, str):
            cleaned_value = float(''.join(filter(lambda x: x.isdigit() or x == '.', row)))
            x_train_cleaned.append(cleaned_value)
        else:
            x_train_cleaned.append(row)
    
    x_train_cleaned = np.array(x_train_cleaned)
    y_train_cleaned = y_train
if isinstance(x_train_cleaned, pd.DataFrame):
    mask = ~(x_train_cleaned.isna().any(axis=1) | pd.isna(y_train_cleaned))
    x_train_cleaned = x_train_cleaned[mask]
    y_train_cleaned = y_train_cleaned[mask]
model.fit(x_train_cleaned, y_train_cleaned)
```
